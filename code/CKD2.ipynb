````python
# Feature Selection Methods

## Correlation Analysis
# Correlation analysis measures the linear relationship between features and the target variable. Features with high absolute correlation with the target are often considered more relevant.
# - **How it works:** Calculates the correlation coefficient (e.g., Pearson, Spearman) between each feature and the target variable.
# - **Potential benefits:** Simple to understand and implement, can identify redundant features.

# ## Mutual Information
# Mutual information measures the dependency between two variables. It quantifies how much information one variable provides about another. For feature selection, it measures the dependency between each feature and the target variable.
# - **How it works:** Based on entropy, it calculates the reduction in uncertainty about the target variable when the value of a feature is known.
# - **Potential benefits:** Can capture non-linear relationships, effective for both numerical and categorical features.

# ## Feature Importance from Tree-Based Models
# Tree-based models like Random Forests and Gradient Boosting can provide a score indicating the importance of each feature in the model's predictions.
# - **How it works:** Importance is typically calculated based on how much each feature contributes to reducing impurity (e.g., Gini impurity, entropy) across all trees in the ensemble.
# - **Potential benefits:** Captures interactions between features, relatively easy to interpret.

# ## Recursive Feature Elimination (RFE)
# RFE is a wrapper method that recursively removes features and builds a model on the remaining features. The performance of the model is used to rank the features.
# - **How it works:** Starts with all features, trains a model, removes the least important feature, and repeats until the desired number of features is reached.
# - **Potential benefits:** Can find optimal subsets of features, considers feature interactions.
````

## Explore cross-validation

### Subtask:
Research and understand cross-validation techniques (e.g., k-fold cross-validation).


**Reasoning**:
The subtask is to research and understand different cross-validation techniques. This does not require any code execution. I will provide a markdown summary of the requested cross-validation techniques.



````python
# Cross-Validation Techniques

# ## K-Fold Cross-Validation
# K-Fold Cross-Validation is a widely used technique to evaluate the performance of a machine learning model on unseen data. It helps in getting a more reliable estimate of model performance compared to a single train-test split, especially with limited data.
# - **How it works:** The dataset is divided into k equal-sized folds. The model is trained on k-1 folds and evaluated on the remaining fold. This process is repeated k times, with each fold serving as the test set exactly once. The final performance metric is the average of the scores obtained in each fold.
# - **Potential benefits:** Provides a more robust estimate of model performance, reduces the variance of the performance estimate compared to a single split, makes better use of the available data.
# - **Potential drawbacks:** Can be computationally expensive for large datasets or complex models.

# ## Stratified K-Fold Cross-Validation
# Stratified K-Fold Cross-Validation is a variation of k-fold cross-validation where each fold contains approximately the same proportion of samples of each target class as the complete set. This is particularly useful for imbalanced datasets to ensure that each fold is representative of the overall data distribution.
# - **How it works:** Similar to k-fold, but the splitting is done in a way that maintains the original class distribution in each fold.
# - **Potential benefits:** Ensures that each fold is representative, important for classification tasks with imbalanced classes.

# ## Leave-One-Out Cross-Validation (LOOCV)
# Leave-One-Out Cross-Validation is an extreme case of k-fold cross-validation where k is equal to the number of data instances. In each iteration, a single data instance is used as the test set, and the remaining instances are used as the training set.
# - **How it works:** The model is trained on n-1 instances and tested on the single remaining instance, repeated n times.
# - **Potential benefits:** Makes maximum use of the data for training, provides a nearly unbiased estimate of the model's performance.
# - **Potential drawbacks:** Very computationally expensive for large datasets.
````

## Implement cross-validation

### Subtask:
Implement k-fold cross-validation on the selected features and the target variable to get a more robust evaluation of the model's performance.


**Reasoning**:
Implement k-fold cross-validation on the selected features and the target variable to get a more robust evaluation of the model's performance by iterating through the models, creating pipelines with a StandardScaler, and using cross_val_score.



````python
from sklearn.model_selection import cross_val_score

# Define the number of folds for cross-validation
n_splits = 10

# Dictionary to store cross-validation results
cv_results = {}

# Iterate through the models
for name, model in models.items():
    # Create a pipeline with StandardScaler and the model
    clf = Pipeline(steps=[("scaler", StandardScaler()), ("classifier", model)])

    # Perform k-fold cross-validation
    scores = cross_val_score(clf, X_selected, y.values.ravel(), cv=n_splits, scoring='accuracy')

    # Store the cross-validation scores
    cv_results[name] = scores

    # Print the cross-validation scores
    print(f"\n{name} (Selected Features) - {n_splits}-Fold Cross-Validation Accuracy:")
    print(f"Scores: {scores}")
    print(f"Mean Accuracy: {scores.mean():.4f}")
    print(f"Standard Deviation: {scores.std():.4f}")
    print("-" * 40)
````

## Analyze cross-validation results

### Subtask:
Analyze the results of the cross-validation to understand the model's performance consistency across different folds.


**Reasoning**:
Access the cross-validation results stored in `cv_results`, calculate the mean and standard deviation for each model's scores, and print these statistics to analyze the performance consistency across folds.



````python
import numpy as np

print("Cross-Validation Results Analysis:")
print("-" * 60)

for name, scores in cv_results.items():
    mean_accuracy = np.mean(scores)
    std_accuracy = np.std(scores)

    print(f"{name}:")
    print(f"  Mean Accuracy: {mean_accuracy:.4f}")
    print(f"  Standard Deviation: {std_accuracy:.4f}")

    if std_accuracy < 0.05:  # Threshold for low standard deviation
        print("  Consistency: High")
    elif std_accuracy < 0.1:
        print("  Consistency: Medium")
    else:
        print("  Consistency: Low")
    print("-" * 60)

print("\nModels with Consistently High Performance (based on mean accuracy and low standard deviation):")
# Identify models with high mean accuracy and low standard deviation
consistent_high_performers = []
for name, scores in cv_results.items():
    mean_accuracy = np.mean(scores)
    std_accuracy = np.std(scores)
    if mean_accuracy > 0.95 and std_accuracy < 0.05: # Example thresholds
        consistent_high_performers.append(name)

if consistent_high_performers:
    print(consistent_high_performers)
else:
    print("No models identified as consistently high performers based on the defined thresholds.")
````

## Consider stratified sampling

### Subtask:
Consider stratified sampling


**Reasoning**:
Re-run the k-fold cross-validation using StratifiedKFold to ensure that the class distribution is preserved in each fold.



````python
from sklearn.model_selection import StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
import numpy as np

# Define the number of folds for stratified cross-validation
n_splits = 10
skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)

# Dictionary to store stratified cross-validation results
stratified_cv_results = {}

# Define the models again to ensure they are in a clean state
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier(),
    "Gradient Boosting": GradientBoostingClassifier(),
    "KNN": KNeighborsClassifier(),
    "SVM": SVC()
}


# Iterate through the models
for name, model in models.items():
    # Create a pipeline with StandardScaler and the model
    clf = Pipeline(steps=[("scaler", StandardScaler()), ("classifier", model)])

    # Perform stratified k-fold cross-validation
    scores = []
    for train_index, test_index in skf.split(X_selected, y.values.ravel()):
        X_train_fold, X_test_fold = X_selected.iloc[train_index], X_selected.iloc[test_index]
        y_train_fold, y_test_fold = y.values.ravel()[train_index], y.values.ravel()[test_index]

        clf.fit(X_train_fold, y_train_fold)
        score = clf.score(X_test_fold, y_test_fold)
        scores.append(score)


    # Store the cross-validation scores
    stratified_cv_results[name] = scores

    # Print the stratified cross-validation scores
    print(f"\n{name} (Selected Features) - {n_splits}-Fold Stratified Cross-Validation Accuracy:")
    print(f"Scores: {scores}")
    print(f"Mean Accuracy: {np.mean(scores):.4f}")
    print(f"Standard Deviation: {np.std(scores):.4f}")
    print("-" * 40)
````
